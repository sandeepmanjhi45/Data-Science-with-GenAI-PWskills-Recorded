Flow control in Python refers to the mechanisms that govern the order in which statements and instructions are executed. This is crucial for implementing logic in your programs.



Flow control decides the **order** in which Python runs our statements.

## The three kinds of flow control
Real-life examples of each kind:
- **Conditional:** I will play cricket *when* the weather is sunny.
- **Loop:** I will keep checking the dishes at a buffet *until* all the dishes are over.
- **Break:** I will check the dishes *until* I find the pasta, and then stop.

This topic covers **conditional statements**. Loops come in the next topic.

## Conditional statements
Conditional statements help us write code that makes decisions based on conditions. There are four forms:
1. `if`
2. `if` ... `else`
3. `if` ... `elif` ... `else`
4. nested `if` ... `else`

## 1. The `if` statement
**Syntax:**
```

    code that runs when the condition is True
```
The code runs only when the condition is `True`. If it is `False`, Python skips it.

### More than one condition
In real life, a decision often depends on several conditions.

````python
#in real life, decisions are made based on multiple conditions

number = 42
if ((number % 2 == 0) & (number > 0)):
    print("This is the number of interest")
````

````python
grade = 8
co_cirr = True
if ((grade >=8) and (co_cirr == True)):
    print("You are all rounder")
````

## 2. The `if` ... `else` statement
If the condition is `True`, the `if` block runs. Otherwise, the `else` block runs.

````python
#if else >> execute the if block when the condition is true else execute else block
weather = "sunny"
if weather == "sunny": #condition is true, execute if block
    print("I will play cricket")
else:#condition is not true, else block will be executed
    print("I will watch TV")
````

## 3. The `if` ... `elif` ... `else` statement
When we have **more than two** choices, we use `elif` (short for "else if").

**Syntax:**
```
if first condition:
    ...
elif second condition:
    ...
else:
    ...
```

````python
a = 100
if a > 100:
    print("This block will be executed if a is greater than 100")
elif a < 100:
    print("This block will be executed if a is lesser than 100")
else:
    print("The number is equals to 100")
````

````python
a = 100000

if a < 100:
    print("This block will be executed if a is lesser than 100")
elif a > 2000:
    print("The number is huge")
elif a > 100:
    print("This block will be executed if a is greater than 100")
else:
    print("The number is equals to 100")
````

`a` is 100000, which is also greater than 100. But Python checks the conditions **from top to bottom** and stops at the **first one that is `True`**. Here `a > 2000` comes first, so the answer is "The number is huge". So the order of the `elif` lines matters.

### A note on `80 <= score < 90`
Python lets us chain comparisons. `80 <= score < 90` means `80 <= score` **and** `score < 90`.

````python
score = 75
if score >= 90:
    print("grade A")
elif (80<=score<90):
    print("Grade B")
elif (70<=score<80):
    print("Grade C")
else:
    print("Grade D")
````

````python
age = 80
if age < 18:
    print("You are minority")
elif 18<age<=65:
    print("You are adult")
else:
    print("You are senior citizen")
````

## 4. Nested `if` ... `else`
We can put an `if` ... `else` **inside** another `if` ... `else`.

````python
#nested if else >> you can have multiple if else inside if else
x = 7
y = 6
if x > 5:
    if y>5:
        print("Both x and y is greater than 5")
    else:
        print("x is greater than 5 but y is less than 5")
else:
    print("x is not greater than 5")
````

````python
is_vip = False
age = 15
if is_vip:
    if age >=18:
        if age < 65:
            print("welcome to vip customer")
        else:
            print("You are vip and very imp. You are eligible for senior citizen discount")
    else:
        print("Vip status is only for adult")
else:
    print("You are not a vip customer and will be charged a regular price")
````

### A small registration check
This uses nested conditions to check a name, an email and a password.

````python
name = input("Please enter your name")
email = input("Please enter your email")
password = input("Please enter your password")

if name == "":
    print("Please enter a valid name, can not be empty")
else:
    if "@" not in email:
        print("Please enter a valid email")
    else:
        if len(password) < 0:
            print("No password entered, pls provide a valid password")
        else:
            print("Login/registration is successfull")
````

**A small bug in the password check:** `len(password) < 0` can never be `True`, because a length is never negative. To catch an empty password, use `len(password) == 0` (or `not password`).

## Summary

| Statement | When we use it |
|-----------|----------------|
| `if` | Run some code only when a condition is `True` |
| `if` ... `else` | Choose between **two** paths |
| `if` ... `elif` ... `else` | Choose between **many** paths |
| nested `if` | Put one decision inside another |

- Put a **colon ( : )** after `if`, `elif` and `else`.
- **Indent** the code inside each block.
- Python checks the conditions from top to bottom and runs the first block whose condition is `True`.
- Use `and`, `or` and `not` to combine conditions.
- `input()` gives text, so use `int()` when you need a number.
