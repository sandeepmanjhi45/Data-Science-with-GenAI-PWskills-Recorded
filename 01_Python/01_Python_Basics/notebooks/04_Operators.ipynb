**What is an operator?** An operator is a special symbol or word that tells Python to do something with values. The values it works on are called **operands**. In `5 + 2`, the `+` is the operator, and `5` and `2` are the operands.

**Why do we need operators?** To work with (manipulate) data.

| # | Type | What it does | Examples |
|---|------|--------------|----------|
| 1 | Arithmetic | Maths | `+  -  *  /  %  **  //` |
| 2 | Comparison | Compares two values, gives `True` or `False` | `==  !=  >  <  >=  <=` |
| 3 | Logical | Combines True/False values | `and  or  not` |
| 4 | Assignment | Stores a value in a variable | `=  +=  -=  *=  /=` |
| 5 | Membership | Checks if something is inside something | `in  not in` |
| 6 | Identity | Checks if two names point to the same object | `is  is not` |
| 7 | Bitwise | Works on the bits (0s and 1s) of numbers | `&  \|  ^  ~  <<  >>` |

## 3. Logical operators
Logical operators combine `True` and `False` values.

| A | B | `A and B` | `A or B` |
|---|---|-----------|----------|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

- `and` is True only when **both** sides are True.
- `or` is True when **at least one** side is True.
- `not` flips the value: `not True` is `False`.

```python
#Logical operator
#and
True and True
```

```python
True and False
```

```python
False and True
```

```python
False and False
```

```python
True or False
```

```python
False or True
```

`True or True` is also `True`. The only way to get `False` from `or` is when both sides are `False`:

```python
False or False
```

## 4. Assignment operators
`=` stores a value in a variable. The shorthand forms (`+=`, `-=`, `*=`, `/=`) change a variable using its own value.

| Shorthand | Same as |
|-----------|---------|
| `a += 5` | `a = a + 5` |
| `a *= 2` | `a = a * 2` |
| `a /= 3` | `a = a / 3` |

```python
#Assignment operator

a  =10
```

```python
a
```

```python
a + 5
```

```python
a
```

Notice that `a` is still 10. The cell `a + 5` only **calculates** the answer. It does not change `a`. To change `a`, we have to store the result back into it, as the next cells do.

```python
a = a + 10
```

## 7. Bitwise operators
Bitwise operators work on the **bits** (the 0s and 1s) of a number.

| Operator | Name | Rule |
|----------|------|------|
| `&` | AND | 1 only if both bits are 1 |
| `\|` | OR | 1 if at least one bit is 1 |
| `^` | XOR | 1 if exactly one bit is 1 |
| `~` | NOT (negation) | flips the bits; `~x` equals `-(x + 1)` |
| `<<` | left shift | adds zeros on the right |
| `>>` | right shift | removes bits from the right |

`bin()` shows a number in binary (the `0b` at the start only means "binary").

```python
#bitwise operator>> operations at bit level
10 & 10
```

```python
18 & 3
```

Let us look at the binary form of the numbers used above:

```python
bin(10)
```

```python
bin(2)
```

```python
3 | 5
```

## Order of precedence
When an expression has many operators, Python follows an order. The top ones run first. Operators at the same level run from **left to right**.

| Order | Operators |
|-------|-----------|
| 1 | `( )` brackets |
| 2 | `**` |
| 3 | `+x`, `-x`, `~x` |
| 4 | `*`, `/`, `//`, `%` |
| 5 | `+`, `-` |
| 6 | `<<`, `>>` |
| 7 | `&` |
| 8 | `^` |
| 9 | `\|` |
| 10 | comparisons: `==`, `!=`, `>`, `<`, `>=`, `<=`, `is`, `in` |
| 11 | `not` |
| 12 | `and` |
| 13 | `or` |

**Tip:** when you are unsure, use brackets. They make the code easier to read, and they always run first.

```python
## order of precedence
#execution will start from left
2+3-1
```

```python
#always parenthesis

(5+3) - 4
```

```python
4 - (5+3)
```

```python
(10+5) + (8/2)
```
