## Implicit and explicit type casting
There are two ways a type can change: Python can do it by itself (**implicit**), or we can do it ourselves (**explicit**).

### Implicit type casting
Python changes the type **by itself**, so we do not have to do anything.

```python
#implicit typecasting >> python understand the datatype
a = 5
type(a)
```

```python
2+3
```

```python
2+3.5
```

In `2 + 3.5`, Python quietly turned the 2 into 2.0 and gave back a float, 5.5. That is implicit type casting.

### Explicit type casting
We change the type **ourselves**, using the built-in functions `int()`, `float()`, `str()` and `bool()`.

```python
#explicit typecasting >> convert data type using inbuilt functiond>> str, int, float, bool
a = "2"
type(a)
```

```python
int(a)
```

```python
float(a)
```

## bool(): True or False
`bool()` turns a value into `True` or `False`. `0` gives `False`, and any other number gives `True`.

```python
#bool
bool(0)
```

```python
bool(1)
```

```python
bool(100)
```

```python
bool(10000)
```

It is not only numbers. Empty values give `False`, and values that contain something give `True`. This cell is my addition:

```python
print(bool(""))
print(bool("Ajay"))
```

## Summary

| Function | Converts to | Example | Result |
|----------|-------------|---------|--------|
| `int()` | whole number | `int("2")` | 2 |
| `int()` | whole number (cuts the decimal) | `int(3.4)` | 3 |
| `float()` | decimal number | `float("2.3")` | 2.3 |
| `float()` | decimal number | `float(2)` | 2.0 |
| `str()` | text | `str(5)` | `'5'` |
| `bool()` | True or False | `bool(0)` | False |
| `bool()` | True or False | `bool(100)` | True |

- **Implicit** type casting: Python changes the type by itself (`2 + 3.5` gives 5.5).
- **Explicit** type casting: we change it ourselves with `int()`, `float()`, `str()` or `bool()`.
- Text that is not a number (like `"Ajay"`) cannot be turned into a number. It gives a `ValueError`.
- Adding text and a number gives a `TypeError`. Convert one of them first.
