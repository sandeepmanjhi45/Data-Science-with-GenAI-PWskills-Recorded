### What are variables (identifiers)?
Variables are placeholders where we keep different kinds of data, so we can use them to write logic.

| Kind of data | Python type | Example |
|--------------|-------------|---------|
| Whole numbers | `int` | `3` |
| Decimal numbers | `float` | `4.5` |
| Text (characters) | `str` | `"Ajay"` |
| True or False | `bool` | `True` |
| No value | `NoneType` | `None` |
| Complex numbers | `complex` | `5+6j` |

```python
b = 3
b
```

**A variable is a reserved space in memory for storing a value.**

```python
b = 4.5
```

```python
b
```

When we give a variable a new value (for example `b = 4.5` after `b = 3`), Python keeps the new value in memory under the same name.

## Rules for naming a variable
- A name must start with a **letter** or an **underscore ( _ )**.
- It can contain letters, digits and underscores, but no spaces or symbols such as `-`, `@` or `%`.
- It must **not be a keyword**.

### Invalid names (these give errors)
- `5 = a` and `2.3 = b`: the left side must be a name, not a number.
- `1name`: a name cannot start with a digit.

## Keywords
Keywords are predefined words that have a special meaning and a fixed purpose in Python, for example `if`, `else`, `for`, `while`, `def`, `True` and `None`. We cannot use them as variable names.

```python
#keywords >> are predefined words that hold a special meaning and have specific purpose in python
```

```python
help('keywords')
```

Another way to see the list of keywords, and to check one word at a time:

```python
import keyword
print(keyword.kwlist)
print(len(keyword.kwlist))
```

```python
print(keyword.iskeyword("if"))
print(keyword.iskeyword("print"))
```

`if` is a keyword, but `print` is not. `print` is a built-in **function**.

## Quick summary
- A variable is a name that holds a value. Its type can be `int`, `float`, `str`, `bool`, `None` or `complex`.
- A variable name starts with a letter or underscore, has no spaces or symbols, and is not a keyword.
- Comments start with `#`. Triple quotes can be used for several lines.
- Keywords are special reserved words. Python is case sensitive.
- `input()` always returns text, and `int()`, `float()`, `str()` and `bool()` change the type.
