## 6. Assignment and copying
### Assignment (`=`): two names, one list
`a = grocery_list` does **not** make a new list. Both names point to the **same** list, so a change made through one name is seen in the other.

````python
#assignment: a and grocery_list are the same list
a = grocery_list
a[0] = "potato"
a
````

````python
grocery_list #the value is also changed in grocery_list
````

### `copy()`: a separate list
`grocery_list.copy()` makes a **new list**. A change in the copy does not change the original list.

````python
b = grocery_list.copy()
b[0] = "Milk"
b
````

````python
grocery_list #not changed by the change in b
````

### Shallow copy and deep copy
`copy()` makes a **shallow copy**: only the outer list is copied. If a list has another list inside it (a nested list), the inner list is still shared. `copy.deepcopy()` makes a **deep copy**, which copies the inner lists too.

````python
import copy
nested = [[1, 2], [3, 4]]
shallow = nested.copy()
deep = copy.deepcopy(nested)
nested[0][0] = 99
shallow #inner list is shared, so 99 shows here too
````

````python
deep #not changed
````

````python
len(book_list) #no of elements in the list
````

````python
book_list.count("Algorithm") #count gives the count of a specific element in a list
````

````python
book_list.sort()
book_list
````

## 9. List comprehension
A list comprehension makes a new list in **one short line**, in place of a `for` loop with `append()`.

Syntax: `[expression for item in list]`

````python
#using a for loop
prices = [10, 20, 30, 40, 50]
doubled_price = []
for i in prices:
    doubled_price.append(i*2)
doubled_price
````

````python
#using list comprehension >> concise code
doubled_price = [price * 2 for price in prices]
doubled_price
````

````python
names = ["ajay", "bijay", "sanjay"]
[name.capitalize() for name in names]
````

````python
#use cases
numbers = [1, 2, 3, 4, 5]
[num**2 for num in numbers]
````

**Use case:** finding the file type (extension) of many files. First for one file, then for the whole list.

````python
str1 = "doc1.ppt"
str1.split(".")[-1]
````

````python
file_name = ["doc1.ppt", "doc2.pdf", "doc3.jpg", "doc4.py"]
[file.split(".")[-1] for file in file_name]
````

### Conditional list comprehension
An `if` at the end keeps only the items that satisfy the condition.

Syntax: `[expression for item in list if condition]`

````python
#conditional list comprehension
email_address = ["aj@gmail.com", "sj@yahoo.com", "rj@yahoo.com", "mj@gmail.com"]
[email for email in email_address if email.endswith("@yahoo.com")]
````

### Nested list comprehension
Two `for` parts in one comprehension work like a loop inside a loop. Here every number of the first list is paired with every number of the second list.

````python
#nested loops
pairs = []
for x in [1, 2, 3]:
    for y in [4, 5, 6]:
        pairs.append([x, y])
pairs
````

````python
#nested list comprehension
[[x, y] for x in [1, 2, 3] for y in [4, 5, 6]]
````

## 10. List as a stack and a queue

### Stack
A stack follows **LIFO**: Last In, First Out. The last item added is the first item removed, and everything happens at **one end** (the top). A pile of plates is a good example: the plate put on top last is taken first.

```
          add / remove here
                 |
             [ plate4 ]   <- top
             [ plate3 ]
             [ plate2 ]
             [ plate1 ]
```

A list works as a stack: `append()` pushes an item on the top and `pop()` removes the top item.

````python
stack_of_plates = []
stack_of_plates.append("plate1")
stack_of_plates.append("plate2")
stack_of_plates.append("plate3")
stack_of_plates.append("plate4")
stack_of_plates
````

````python
stack_of_plates.pop()
````

````python
stack_of_plates
````

**Use case:** browser history. The back button opens the page visited last.

````python
#use case
browsing_history = []
browsing_history.append("home_page")
browsing_history.append("about us")
browsing_history.append("contact")
browsing_history.pop()
````

````python
browsing_history
````

### Queue
A queue follows **FIFO**: First In, First Out. Items are added at one end (the rear) and removed from the other end (the front), in an orderly sequence. A line of customers at a checkout counter is a good example.

```
 remove here                                   add here
 (front)                                        (rear)
    <-  [ cus1 | cus2 | cus3 | cus4 | cus5 ]  <-
```

`deque` is used for a queue because it can remove items from the front quickly. `collections` is a module (a collection of ready-made Python code) and `deque` is a class inside it.

````python
from collections import deque
````

````python
checkout = deque()
checkout.append("cus1")
checkout.append("cus2")
checkout.append("cus3")
checkout.append("cus4")
checkout.append("cus5")
checkout
````

````python
while checkout:
    customer = checkout.popleft() #removes from the front
    print("serving", customer)
````

The `queue` module also has a `Queue` class. `put()` adds an item at the rear and `get()` removes the item from the front.

````python
from queue import Queue
print_queue = Queue()
print_queue.put("Print the 1 pdf") #put adds an item at the rear
print_queue.put("Print the 2 pdf")
print_queue.put("Print the 3 pdf")
print_queue.put("Print the 4 pdf")
print_queue.put("Print the 5 pdf")


while not print_queue.empty():
    print_job = print_queue.get() #get removes the item from the front
    print("printing", print_job)
````

| Task | Code | Result |

| Add one item at the end | `lis.append("Banana")` | list gets longer |

**Key learnings**
- A list is ordered, can hold any data type, and is mutable.
- Slicing and indexing work like strings. The end index of a slice is not included.
- `a = b` does not copy a list. `copy()` makes a shallow copy and `copy.deepcopy()` makes a deep copy.
- `sorted()` gives a new list, while `sort()` changes the same list.
- A list comprehension is a short way to write a `for` loop that builds a list.
- Stack is LIFO (use `append()` and `pop()`). Queue is FIFO (use `deque` with `popleft()`).
