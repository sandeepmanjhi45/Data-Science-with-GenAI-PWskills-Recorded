## 1. How a computer stores characters

A computer understands only **numbers**. So every character (letter, digit, symbol) is given a number, called its **code**.

- **ASCII** (American Standard Code for Information Interchange) uses **7 bits**, so it can store 2^7 = **128** characters. It covers English letters, digits and common symbols.
- **Unicode** is a much bigger system. It gives a number to characters of almost every language, plus symbols and emojis.
- **UTF** (Unicode Transformation Format) is the way Unicode characters are saved. **UTF-8** and **UTF-16** are the popular ones, and UTF-8 is the most widely used in industry.

### ASCII: `ord()` and `chr()`
- `ord(character)` gives the **number** of a character (ord means *ordinal*, the numerical order of the character).
- `chr(number)` gives the **character** for a number.

## 4. Indexing and slicing
### Indexing: reading one character
Every character has a position called an **index**.
- From the left, the first index is **0**.
- From the right, we use negative numbers, so **-1** is the last character.

| Character | A | j | a | y |
|-----------|---|---|---|---|
| Index from left | 0 | 1 | 2 | 3 |
| Index from right | -4 | -3 | -2 | -1 |

```python
#Extract/slicing of string
a = "Ajay"
a[0]
```

```python
a[1]
```

```python
a[-1]
```

```python
a[3]
```

`"Ajay"` has indexes 0 to 3 only, so index 4 gives an `IndexError`.

```python
a[4] #it will throw an error as index 4 doesn't exist
```

### Slicing: reading a part of a string
Syntax: `string[start:end]`
- The **start** index is included.
- The **end** index is **not** included, so the result stops at `end - 1`.
- If we leave out the end, Python goes up to the last character.

```python
string1 = "I am a good student"
string1[0:4] #4 is exclusive [a, b) >> it will give result till b-1 index
```

```python
string1[5:12]
```

```python
string1[0:] #if you do not provide the end index, it gives the substring till the last index
```

```python
string1[3:]
```

Slicing also works with negative indexes.

```python
strings2 = "pwskills"
```

```python
strings2[1]
```

```python
strings2[-1]
```

```python
strings2[-2]
```

```python
#except the last three characters
strings2[:-3]
```

```python
#only the last three characters
strings2[-3:]
```

### Step in slicing
Full syntax: `string[start:end:step]`
- The default step is **1** (take every character).
- Step **2** takes every second character.
- A **negative step** moves backwards, so `[::-1]` reverses the whole string.

```python
string2 = "hello world!"
```

```python
string2[0:5:1] #by default step is 1
```

```python
string2[0:5:2] #syntax >> [start:end:step]
```

```python
string2[::1]
```

```python
string2[::2]
```

```python
#iterate over the sequence in reverse order
#it means start from the end, move to the beginning and step backwards by one character
string2[::-1]
```

Changing a character gives a `TypeError`, because a string does not support item assignment.

```python
s[0] = "a"
```

```python
s
```

To get a changed string, build a **new** one from the old one:

```python
"a" + s[1:]
```

## 7. Changing text with string methods
### `replace()`
`replace(old, new)` swaps every `old` text with the `new` text.

```python
#modification of string
s = "I am a student"
s.replace("student", "teacher")
```

`replace()` does **not** change `s`. Strings are immutable, so it returns a **new** string. The original stays the same:

```python
s
```

**Use case:** correcting an address.

```python
#use case
address = "123 suhana apt, marathalli, Bangalore"
address
```

```python
address.replace("apt", "apartment")
```

**Use case:** expanding short chat words. We can chain many `replace()` calls one after another.

```python
text = "BRB, TTYL, LOL"
text.replace("BRB", "Be right back").replace("TTYL","talk to you later").replace("LOL", "Laughing out loud")
```

```python
text = "Hello World"
text.lower()
```

```python
text.upper()
```

```python
text.title()
```

```python
text = "I am a Student"
text.swapcase()
```

```python
text.capitalize()
```

## 8. Searching in a string
The `in` keyword checks whether a smaller text is present inside a bigger text. The answer is `True` or `False`.

```python
#string searching
sentence = "I am a student at pwskills"
"am" in sentence
```

```python
sentence = "I am a student at pwskills"
search_word = "mat"
if search_word in sentence:
    print("The search word is present")
else:
    print("Not present")
```

**Use case:** a simple email check (does it contain `@`?).

```python
email = "ajay123@gmail.com"
if "@" in email:
    print("valid email")
else:
    print("Invalid email")
```

**Use case:** checking a file type.

```python
document = "some_file.pdf"
if ".pdf" in document:
    print("Its a pdf file")
else:
    print("Not a pdf file")
```

## 9. Comparing strings
`==` checks if two strings are exactly equal. Capital and small letters are **different**, so `"Hello"` and `"hello"` are not equal. To ignore the case, convert both to lower case first.

```python
#string comparison
str1 = "hello"
str2 = "hello"
str1 == str2
```

```python
str1 = "Hello"
str2 = "hello"
str1 == str2
```

```python
str1 = "Hello"
str2 = "hello"
str1.lower() == str2.lower()
```

**Use case:** is a username already taken? Here `in` checks the whole items of the list.

```python
reg_username = ["pw", "PW", "pW"]
new_user = "PW1"
if new_user in reg_username:
    print("already in use")
else:
    print("username is available")
```

**Use case:** matching a scanned product code. The program asks for the code and compares it with the correct one.

```python
product_code = "P123"
scanned_code = input("scan the product code ")
if product_code == scanned_code:
    print("The price is 100")
else:
    print("Talk to receptionist")
```

## 12. Escape sequences
An **escape sequence** is a special combination of characters inside a string. It starts with a backslash `\`.

| Escape | Meaning |
|--------|---------|
| `\n` | new line |
| `\t` | tab (a wide space) |
| `\\` | one backslash |
| `\'` | single quote |
| `\r` | carriage return (moves the cursor back to the start of the line) |

Tip: to see the effect of an escape sequence, use `print()`.

```python
address = """123, sanmukha villa,
marathalli,
bangalore"""
```

```python
address
```

```python
print(address)
```

```python
address = """123, sanmukha villa,\nmarathalli, \nbangalore"""
```

```python
print(address) #\n >> to insert a new line character
```

```python
#\t escape character \t >> tab
table = "Name\tAge\tGrade\nAjay\t21\tA\nBijay\t22\tA+"
print(table)
```

```python
file_path = "C:\\USERS\\DOWNLOADS\\FILES.txt"
print(file_path)
```

```python
#\' single quote escape
str1 = 'I can\'t believe its friday'
print(str1)
```

```python
#carriage return \r >> moves the cursor to the beginning of the line
print("Hello, I am \rAjay, I am going to delhi")
```

## 13. String formatting
**String formatting** means making a string with **placeholders**, which are filled with values of variables. Without formatting, we would have to write a new line for every name:

```python
#string formatting >> creating a string with placeholders for variables
print("My name is Ajay")
print("My name is Bijay")
print("My name is Sanjay")
```

There are three ways to format a string:
1. **f-string** (formatted string literal): put `f` before the quotes and write the variable inside `{ }`. This is the easiest and most used way today.
2. **`%` formatting**: the older way (`%s` is a placeholder for a string).
3. **`.format()`**: put `{}` placeholders in the text and fill them in order.

### f-string

```python
name = "Bijay"
greetings = f"hello {name}, how are you?"
print(greetings)
```

```python
#f string (formatted string literals)
course = "pw data + frontend"
duration = 4
f"My course is {course} and its duration is {duration} years."
```

```python
user_name = "Ajay"
post_date = "2024-06-03"

print(f"Hello user, {user_name}, Today is {post_date} post.")
```

We can also do a calculation inside `{ }`. Here we change Celsius to Fahrenheit.

```python
celsius = 30
f"{celsius} is equals to {(celsius * 9/5)+32}F"
```

### `%` formatting

```python
#string formatting
name = "PW Skills"
greeting = "Hello %s, wish you a wonderful day"%name
print(greeting)
```

### `.format()`

```python
#.format
template = "Hello, {} welcome to {}"
message = template.format("Ajay", "Data Science")
print(message)
```

## 14. More useful checks
- `startswith(text)` is `True` if the string begins with the given text.
- `endswith(text)` is `True` if the string ends with the given text.
- `isalnum()` is `True` if the string has **only letters and digits** (alphanumeric).

```python
str1 = "pwskills"
str1.startswith("p")
```

```python
str1.startswith("pws")
```

```python
str1.startswith("aws")
```

```python
str1.endswith("lls")
```

```python
str1.endswith("lly")
```

```python
a = "123abc"
a.isalnum()
```

```python
a = "Abs"
a.isalnum()
```

```python
a = "123"
a.isalnum()
```

A space or a symbol is **not** a letter or a digit, so `isalnum()` becomes `False`:

```python
a = "abc 123"
a.isalnum()
```

## 15. Looping through a string
A string is a sequence, so a `for` loop can visit **each character** one by one. Here we also count the characters. The count is the same as `len()`.

```python
s = "pwskills"
len(s)
```

```python
count = 0
for i in s:
    print(i)
    count = count+1
print(count)
```

## Summary

| Task | Code | Result |
|------|------|--------|
| Join strings | `"Hello" + " world"` | `'Hello world'` |
| Get one character | `"Ajay"[0]` and `"Ajay"[-1]` | `'A'` and `'y'` |
| Slice (start included, end excluded) | `"pwskills"[0:3]` | `'pws'` |
| Slice with a step | `"hello world!"[::2]` | `'hlowrd'` |
| Reverse | `"hello"[::-1]` | `'olleh'` |
| Count characters | `len("pwskills")` | `8` |
| Replace text | `"I am a student".replace("student", "teacher")` | `'I am a teacher'` |
| Change case | `lower()`, `upper()`, `title()`, `swapcase()`, `capitalize()` | |
| Search | `"am" in "I am a student"` | `True` |
| Remove outer spaces | `"  Apple  ".strip()` | `'Apple'` |
| Cut into a list | `"a,b,c".split(",")` | `['a', 'b', 'c']` |
| Start or end check | `startswith()`, `endswith()` | `True` or `False` |
| Letters and digits only | `"abc123".isalnum()` | `True` |
| f-string | `f"Hello {name}"` | text with the value filled in |

**Key learnings**
- A string is a sequence of characters, and every character has a number (ASCII or Unicode).
- Indexes start from 0 on the left and from -1 on the right. In a slice, the end index is not included.
- Strings are **immutable**. Methods like `replace()` and `strip()` give a **new** string and do not change the old one.
- `+` joins strings, but `-` does not work on them.
- Capital and small letters are different, so use `lower()` to compare without caring about the case.
- Escape sequences such as `\n` and `\t` start with a backslash. Use `print()` to see their effect.
- f-strings are the simplest way to put variables inside text.
