# Part A: Tuples

An item of a tuple cannot be changed after the tuple is created. Because of this, tuples are used for data that should stay safe from changes, such as a list of employee ids.

# Part B: Sets
## 4. Creating a set
- A set is an **unordered** collection of **unique** elements.
- It does not allow duplicate elements.
- It is unordered, so indexing does not work.
- Empty curly brackets `{}` make a **dictionary**, not a set. An empty set is made with `set()`.

```python
s = {}
type(s) #{} is a dictionary
```

```python
type(set()) #an empty set
```

```python
s = {1, 2, 3, 4}
type(s)
```

## 5. Sets remove duplicates
Duplicate items are kept only once. Capital and small letters are different, so `"Ajay"` and `"ajay"` are two items.

```python
s = {1, 2, 1, 2, 2, "Ajay", "Ajay", "Ajay", "ajay", "ajay"}
s
```

**Use case:** removing the duplicates of a list by changing it into a set and back into a list. The order of items in a set is not fixed.

```python
#use case
list1 = [1, 2, 3, "Brinjal", "Brinjal", "Apple", "Apple"]
s = set(list1)
s
```

```python
list1 = list(s)
list1
```

## 6. No indexing in sets
A set is unordered, so its items have no position. Indexing and slicing give a `TypeError`. A `for` loop can still visit every item.

```python
s[0] #it will give error, as a set is not indexed
```

```python
for i in s:
    print(i)
```

## 7. Items of a set must be hashable
A set can store only **hashable** items, which are items with a fixed hash value. Immutable types such as numbers, strings and tuples are hashable. Mutable types such as lists and sets are not hashable, so they cannot be items of a set.

```python
s = {1, 2, 3, 4, 5, {4, 5, 6}} #it will give error, as a set is mutable (unhashable type)
```

```python
{1, 2, 3, 4, 5, [4, 5, 6]} #it will give error, as a list is mutable (unhashable type)
```

```python
{1, 2, 3, 4, 5, (4, 5, 6)} #no error, as a tuple is immutable (stable hash value)
```

## 8. Changing a set
A set is mutable, so items can be added and removed.

| Method | What it does |
|--------|--------------|
| `add(item)` | adds one item |
| `update(iterable)` | adds many items from a list, string, tuple or another set |
| `pop()` | removes one item, which is not necessarily the last |
| `remove(item)` | removes a specific item, and gives an error if it is missing |
| `discard(item)` | removes a specific item, and gives no error if it is missing |
| `clear()` | removes all items, but the empty set stays |

```python
#set is mutable
s = {1, 2, 3, 4}
s.add(100)
s
```

```python
s.add(5) #it is not compulsory that it is added at the end of the set
s
```

```python
s.pop() #it is not compulsory that it removes the last element
```

```python
s.remove(3) #removes a specific element
s
```

```python
s.pop(4) #it will give error, as pop() takes no argument
```

### `update()`

```python
s.update("Ajay") #update takes an iterable, so each character is added
s
```

```python
s.update(["Ajay"]) #a list with one string adds the whole word
s
```

```python
s.update([2, 3]) #2 is already present, so only 3 is added
s
```

### `clear()`

```python
s = {(1, 2, 3), 100, '2', 3.4} #a set can hold different hashable data types
s.clear() #removes all the elements, the empty set stays
s
```

### `remove()` and `discard()`

```python
s = {(1, 2, 3), 100, 101, 102, '2', 3.4}
s.remove(101)
s
```

```python
s.remove(110) #it will give error, as 110 is not present
```

```python
s.discard(100)
s
```

```python
s.discard(110) #error will not be thrown unlike remove(), if the element is not present
s
```

## 9. Set operations
| Operation | Symbol | Method | Meaning |
|-----------|--------|--------|---------|
| Union | `\|` | `union()` | all the elements of both sets, without duplicates |
| Intersection | `&` | `intersection()` | only the common elements |
| Difference | `-` | `difference()` | elements in the first set and not in the second |
| Symmetric difference | `^` | `symmetric_difference()` | elements in either set, but not in both |

```python
s1 = {"hiking", "reading", "coding"}
s2 = {"coding", "photography", "travelling"}
print(s1)
print(s2)
```

```python
#union >> combine all the elements
s1 | s2
```

```python
#intersection >> common elements
s1 & s2
```

```python
#difference
s1 - s2
```

```python
#symmetric difference
s1 ^ s2
```

## 10. Frozen sets
A **frozenset** is the immutable version of a set. Items cannot be added or removed after it is created.

```python
#a normal set can be changed
s = {1, 2, 3}
s.add(100)
s
```

```python
#frozenset >> immutable version of set
my_fs = frozenset([1, 2, 2, 2, 3, 3, 4, 5])
my_fs
```

```python
type(my_fs)
```

```python
my_fs.add(5) #it will give error, as a frozenset cannot be changed
```

## Summary

| | List | Tuple | Set | Frozenset |
|---|------|-------|-----|-----------|
| Brackets | `[ ]` | `( )` | `{ }` | `frozenset([ ])` |
| Ordered | Yes | Yes | No | No |
| Indexing and slicing | Yes | Yes | No | No |
| Mutable | Yes | No | Yes | No |
| Duplicates allowed | Yes | Yes | No | No |

**Key learnings**
- A tuple is like a list that cannot be changed. It is used for data that should stay the same.
- `count()` and `index()` are the two tuple methods. `+` and `*` create a new tuple.
- `{}` is an empty dictionary. An empty set is `set()`.
- A set keeps only unique items and has no order, so it has no indexing.
- Only hashable (immutable) items can be stored in a set. A tuple is allowed, a list or a set is not.
- `remove()` gives an error for a missing item, while `discard()` does not. `pop()` removes an arbitrary item.
- Set operations: union `|`, intersection `&`, difference `-`, symmetric difference `^`.
- A frozenset is a set that cannot be changed.
