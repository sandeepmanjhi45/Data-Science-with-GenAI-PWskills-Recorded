**What is a dictionary?** A dictionary stores data as **key-value pairs** inside curly brackets `{ }`. A value is found by using its key, like finding the meaning of a word in a real dictionary.

- Keys are **unique** and **immutable** (hashable).
- Values can be of any data type.
- A dictionary keeps the **order in which the items were added** (this is guaranteed from Python 3.7). Items are read with a key, not with a position.

**In this notebook:** creating a dictionary, accessing and changing values, rules for keys, adding and deleting items, dictionary methods, dictionary comprehension with `zip()`, and nested dictionaries.

## 1. Creating a dictionary

```python
d = {}
type(d)
```

```python
d = {"name":"Ajay", "email":"aj@gmail.com", "contact": 1234}
d
```

## 2. Accessing and changing values
A list is read with a position (index). A dictionary is read with a **key**.

```python
l = [1, 2, 3, 4]
l[0] #a list is read by position
```

```python
d['name']
```

```python
d["email"]
```

A value can be read only with its key. It is not possible the other way round, from a value to its key.

```python
d[1] #it will give error, as 1 is not a key of d
```

A dictionary is mutable, so the value of a key can be changed.

```python
d['name'] = "Bikay"
d
```

## 3. Rules for keys
A key must be **unique** and **immutable** (hashable).
- Allowed: strings, integers, floats, booleans and tuples.
- Not allowed: lists, sets and dictionaries, because they are mutable.

```python
#rules of defining keys
d = {1:"abc"} #integer as key
d
```

```python
d = {1.2: "abc"} #float as key
d
```

```python
d = {True: "any value"} #boolean value can also be a key of a dictionary
d
```

```python
d = {(1, 2, 3): "abc"} #tuple as key
d
```

```python
d = {[1, 2, 3]: "shyam"} #it will give error, as a list is mutable (unhashable type)
```

```python
d = {{"key":123} : "ajay"} #it will give error, as a dictionary is mutable (unhashable type)
```

A symbol such as `@` or `#` can be a key only when it is written as a **string** inside quotes. Without quotes it is a syntax error.

```python
d = {@: "ram"} #it will give error, as @ without quotes is not valid
```

```python
d = {"@": "ram", "#": 123} #symbols as strings work as keys
d
```

### Repeated keys
If a key is repeated, the last value is kept.

```python
d = {"name":"Ajay", "email":"aj@gmail.com", "contact": 1234, "name": "Bijay"}
d #always the last key, value will be used
```

## 4. Values can be of any type
A value can be a string, a number, a list, a tuple or even another dictionary.

```python
d = {'name': ['Sanjay', "Ajay", 1, 2], 'email': 'aj@gmail.com', 'contact': (1234, 678)}
d
```

```python
d['name'][1] #second item of the list stored in 'name'
```

## 5. Adding and deleting items
- `d[key] = value` adds a new pair, or changes the value if the key already exists.
- `del d[key]` deletes one pair.
- `d.clear()` deletes all the pairs, but the empty dictionary stays.
- `del d` deletes the whole dictionary variable.

```python
d["address"] = "Bangalore" #to create a new key value pair
d
```

```python
del d["address"] #to delete a key value pair
d
```

```python
del d #delete the whole dictionary
```

```python
d #it will give error, as the dictionary is deleted
```

```python
d = {'name': ['Sanjay', "Ajay", 1, 2], 'email': 'aj@gmail.com', 'contact': (1234, 678)}
d.clear() #empty dictionary
d
```

## 6. Dictionary methods
| Method | What it does |
|--------|--------------|
| `len(d)` | number of key-value pairs |
| `keys()` | all the keys |
| `values()` | all the values |
| `items()` | all the (key, value) pairs |
| `copy()` | a new dictionary with the same pairs |
| `pop(key)` | removes the key and gives back its value |
| `update(other)` | adds the pairs of another dictionary |
| `get(key)` | gives the value, and gives `None` instead of an error if the key is missing |
| `fromkeys(keys, value)` | makes a dictionary where every key gets the same value |

`keys()`, `values()` and `items()` give **view objects**, which show the current content of the dictionary.

```python
d = {'name': ['Sanjay', "Ajay", 1, 2], 'email': 'aj@gmail.com', 'contact': (1234, 678)}
len(d)
```

```python
d.keys()
```

```python
d.values()
```

```python
d.items()
```

### `copy()`

```python
d1 = d.copy() #a separate dictionary with the same pairs (shallow copy)
d1
```

```python
del d1 #deleting the copy does not change d
d
```

### `pop()`

```python
d.pop("contact") #removes the key and gives back its value
```

```python
d
```

### `update()`

```python
d1 = {"course": "ds"}
d.update(d1) #adds the key value pairs of d1 into d
d
```

### `get()`
`d[key]` and `d.get(key)` give the same value when the key is present. When the key is missing, `d[key]` gives a `KeyError`, but `get()` gives `None`, or a default value if one is given.

```python
d["name"]
```

```python
d.get("name")
```

```python
print(d.get("phone")) #no error, as the key is missing it gives None
print(d.get("phone", "not available")) #a default value can be given
```

### `fromkeys()`
Every key gets the **same** value, here the whole tuple `("a", "b", "c")`.

```python
#from keys
d.fromkeys((1, 2, 3), ("a", "b", "c"))
```

## 7. Dictionary comprehension
`zip()` pairs the items of two lists one by one. This is a quick way to build a dictionary from two lists.

```python
#zip function
students = ["Arun", "Ajay", "Bijay"]
marks = [80, 95, 100]

for i in zip(students, marks):
    print(i)
```

Storing the above information as key-value pairs in a dictionary, first with a `for` loop:

```python
students_marks = {}
for student, mark in zip(students, marks):
    students_marks[student] = mark
students_marks
```

The same result with a **dictionary comprehension**, in one line. Syntax: `{key: value for key, value in iterable}`

```python
#using dictionary comprehension method
{student: mark for student, mark in zip(students, marks)}
```

**Use case:** joining user ids with user names.

```python
#use case
user_id = [1, 2, 3]
user_name = ["a21", "a22", "a23"]
{u_id: u_name for u_id, u_name in zip(user_id, user_name)}
```

## 8. Nested dictionaries
A value of a dictionary can be another dictionary. The first key finds the inner dictionary, and the second key finds a value inside it.

```python
employee_records = {
    'E001': {'name': 'John Doe', 'department': 'Engineering', 'salary': 75000},
    'E002': {'name': 'Jane Smith', 'department': 'Marketing', 'salary': 65000},
    'E003': {'name': 'David Lee', 'department': 'Finance', 'salary': 80000}
}
employee_records["E001"]
```

```python
employee_records["E001"]["salary"] #first key, then the key inside it
```

Another example, a restaurant menu:

```python
restaurant_menu = {
    'Dish1': {'name': 'Pasta Carbonara', 'price': 15.99, 'description': 'Creamy pasta with bacon and parmesan'},
    'Dish2': {'name': 'Chicken Caesar Salad', 'price': 12.50, 'description': 'Grilled chicken with romaine lettuce and Caesar dressing'},
    'Dish3': {'name': 'Margherita Pizza', 'price': 14.00, 'description': 'Pizza topped with tomato, mozzarella, and basil'}
}
restaurant_menu["Dish1"]
```

`items()` gives the key and the value together, so a `for` loop can go through every dish.

```python
for dish, details in restaurant_menu.items():
    print(dish, "-", details["name"], "-", details["price"])
```

**Key learnings**
- A dictionary stores key-value pairs. A value is read with its key, not with a position.
- Keys must be unique and immutable. A tuple can be a key, but a list, a set or a dictionary cannot.
- A repeated key keeps the last value.
- `d[key]` gives a `KeyError` for a missing key, while `d.get(key)` gives `None`.
- `copy()` makes a shallow copy, as in lists.
- `zip()` with a dictionary comprehension builds a dictionary from two lists.
- Dictionaries can be nested, and `items()` makes it easy to loop through them.
