## 1. What is a programming language?

A **programming language** is a formal language used to give instructions to a computer. It is a set of rules and symbols that define how we write computer programs.

**Compare it with English:**

| English | Programming language |
|---------|----------------------|
| Alphabets: A, B ... Z | Letters, digits and symbols |
| Words: apple, ball | Keywords and names |
| Sentence: "I eat something" | A line of code (a statement) |
| Grammar and tenses | Syntax (the rules of the language) |

In short, a programming language is a set of special rules and symbols that you use to give instructions to a computer.

## 6. Where do we write and run code? (IDE)

Code needs a place to be written and run. That place is called a **platform** or **environment**.
