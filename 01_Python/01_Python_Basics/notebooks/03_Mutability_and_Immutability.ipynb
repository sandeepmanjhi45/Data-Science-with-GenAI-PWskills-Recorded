## Immutability
### Strings are immutable
We can read a character of a string, but we cannot change it. The next two cells give an error on purpose.

```python
b = "pwskills"
b
```

```python
b[0] = "I"
```

```python
b[5] = "Ram"
```

The error says `'str' object does not support item assignment`. In other words, a string does not let us change one of its characters.

**Wait, we did change `b` earlier (`b = "Ajay"` and then `b = "pwskills"`). Is that not a change?**

No. Giving `b` a new value only makes the name `b` point to a **new** string. The old string itself was never changed. What is not allowed is changing a character *inside* a string.

**How do we get a changed string?** We build a new one from the old one. Here we make a new string that starts with `"I"` and keeps the rest of `b`:

```python
"I" + b[1:]
```

### What does immutable mean?

```python
#objects/container whose state or value can not be changed after they are created are
#called as immutable objects or container,
#string is a type of immutable object/ doesn't supports item assignment
```

## Summary

| | Mutable | Immutable |
|---|---------|-----------|
| Can it be changed in place? | Yes | No |
| Examples | `list` (also `dict` and `set`) | `str`, `int`, `float`, `bool`, `tuple` |
| `x[0] = value` | Works | Gives a `TypeError` |

- **Mutable** means the value can be changed after it is created. A list is mutable.
- **Immutable** means the value cannot be changed after it is created. A string is immutable.
- Negative indexes count from the end, so `b[-1]` is the last character.
